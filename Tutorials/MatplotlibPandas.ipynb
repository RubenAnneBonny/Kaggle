### Load Fifa Data

```python
fifa = pd.read_csv("datasets/fifa_data.csv")

fifa.head(5)
```

### Historgrams

```python
bins = []
start = 40
while start <= 100:
    bins.append(start)
    start += 10

plt.hist(fifa.Overall, bins = bins, color="#abcdef")

plt.xticks(bins)

plt.ylabel("Number of Players")
plt.xlabel("Skill level")
plt.title("Distrubution of Player Skils in FIFA 2018")

plt.show()
```

### Pie Charts

```python
left = fifa.loc[fifa["Preferred Foot"] == "Left"].count()["ID"]
right = fifa.loc[fifa["Preferred Foot"] == "Right"].count()["ID"]

left, right
```

```python
labels = ["Left", "Right"]
colors = ["#f03f6b", "#11cf21"]

plt.pie([left, right], labels=labels, colors=colors, autopct="%.2f %%")

plt.title("Foot Preference of FIFA Players")

plt.show()
```

```python
fifa.Weight
```

```python
fifa.Weight = [int(x.strip("lbs")) if type(x) == str else x for x in fifa.Weight]

fifa.Weight
```

```python
light = fifa.loc[fifa.Weight < 125].count()["ID"]
light_medium = fifa.loc[(fifa.Weight >= 125) & (fifa.Weight < 150)].count()["ID"]
medium = fifa.loc[(fifa.Weight >= 150) & (fifa.Weight < 175)].count()["ID"]
medium_heavy = fifa.loc[(fifa.Weight >= 175) & (fifa.Weight <  200)].count()["ID"]
heavy = fifa.loc[fifa.Weight >= 200].count()["ID"]

light, light_medium, medium, medium_heavy, heavy
```

```python
weights = [light, light_medium, medium, medium_heavy, heavy]
labels = ["<125", "125-150", "150-175", "175-200", ">200"]
explode = (.65, .1, 0, 0, .65)

plt.style.use("ggplot")

plt.pie(weights, labels=labels, autopct="%.2f %%", pctdistance=0.8, explode=explode)

plt.title("Weight Distrubution of FIFA Players (in lbs)")

plt.show()
```

### Box Plot

```python
barcelona = fifa.loc[fifa.Club == "FC Barcelona"]["Overall"]
madrid = fifa.loc[fifa.Club == "Real Madrid"]["Overall"]
revs = fifa.loc[fifa.Club == "New England Revolution"]["Overall"]

labels = ["FC Barcelona", "Real Madrid", "New England Revolution"]

plt.style.use("default")

plt.figure(figsize=(5, 8))

boxes = plt.boxplot([barcelona, madrid, revs], labels = labels, patch_artist=True, medianprops={"linewidth": 2})

colors = ["#abcdef", "#1b0b63", "#14fd32"]
ind = 0 
for box in boxes["boxes"]:
    # Set edge color
    box.set(color=colors[ind], linewidth=2)

    # Change fill color
    box.set(facecolor=colors[(ind + 1) % len(colors)])

    ind += 1

plt.title("Professional Soccer Team Comparision")
plt.ylabel("Fifa Overall Rating")

plt.show()
```
